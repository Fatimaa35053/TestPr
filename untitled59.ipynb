```python
import pandas as pd

# Load our student retention dataset
df = pd.read_csv("/content/university_student_retention_dataset_2134 (1).csv")

# Show the first five student records
print("First five rows:")
display(df.head())

# Check the dataset size
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# Check column names
print("Dataset columns:")
print(df.columns.tolist())

# Check the data types
print("Dataset information:")
df.info()

# Check missing values
print("Missing values in each column:")
print(df.isnull().sum())

# Check total missing values
print("Total missing values:", df.isnull().sum().sum())

# Check duplicate rows
print("Duplicate records:", df.duplicated().sum())

# Check the dropout target
print("Student dropout distribution:")
print(df["next_semester_dropout"].value_counts())
```

```python
# STAGE 3 - DATA UNDERSTANDING AND CLEANING

# We already checked duplicate rows in Stage 2.
# Now we check if any student IDs are repeated.

print("Duplicate student IDs:",
      df["student_id"].duplicated().sum())
```

```python
# Let's check if our numerical values make sense.

numeric_columns = [
    "age",
    "gpa_current",
    "gpa_history",
    "attendance_rate",
    "enrollment_gap_months",
    "credits_completed",
    "academic_warning_count",
    "advisor_meeting_count"
]

# Look at the minimum, maximum, and average values.
display(df[numeric_columns].describe())

# Check GPA and attendance.
print("Invalid current GPA:",
      (~df["gpa_current"].between(0, 4)).sum())

print("Invalid previous GPA:",
      (~df["gpa_history"].between(0, 4)).sum())

print("Invalid attendance:",
      (~df["attendance_rate"].between(0, 1)).sum())

# Check for negative values in count columns.
for column in [
    "enrollment_gap_months",
    "credits_completed",
    "academic_warning_count",
    "advisor_meeting_count"
]:
    print(column, "negative values:",
          (df[column] < 0).sum())
```

```python
# We use the IQR method to find unusual values.
# We will review them instead of removing them automatically.

for column in numeric_columns:

    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)

    iqr = q3 - q1

    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr

    outliers = (
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ).sum()

    print(column, "- possible outliers:", outliers)
```

```python
# First, we make a copy of our original dataset.
# This way, the original data stays unchanged.

clean_df = df.copy()

# Remove columns that are not suitable for prediction.
# Some contain personal information or possible data leakage.

columns_to_remove = [
    "student_id",
    "dropout_risk",
    "dropout_type",
    "dropout_duration",
    "dropout_cause",
    "counseling_notes",
    "advising_notes",
    "gpa_by_semester",
    "attendance_by_semester",
    "gender",
    "ethnicity"
]

clean_df = clean_df.drop(columns=columns_to_remove)

print("Remaining columns:")
print(clean_df.columns.tolist())
```

```python
# We create a new column to see how a student's GPA changed.

clean_df["gpa_change"] = (
    clean_df["gpa_current"] - clean_df["gpa_history"]
)

# Let's check our new column.
display(
    clean_df[["gpa_current", "gpa_history", "gpa_change"]].head()
)
```

```python
# Change financial aid status from True/False to 1/0.
clean_df["financial_aid_status"] = (
    clean_df["financial_aid_status"].astype(int)
)

# Make sure the dropout target is numerical.
# 1 = Dropped out, 0 = Continued studying.

clean_df["next_semester_dropout"] = (
    clean_df["next_semester_dropout"].astype(int)
)

print("Updated data types:")
print(clean_df.dtypes)
```

```python
# Let's check the dataset after our changes.

print("Cleaned dataset size:", clean_df.shape)

print("Missing values left:",
      clean_df.isnull().sum().sum())

print("Duplicate rows left:",
      clean_df.duplicated().sum())

# Look at the first five rows.
display(clean_df.head())
```

```python
# Save our cleaned dataset as a new CSV file.

clean_df.to_csv("student_retention_cleaned.csv", index=False)

print("Our cleaned dataset has been saved!")
```

### Stage 5 – Model Development and Evaluation

```python
# 5.1 Import Machine Learning Libraries

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report,
    RocCurveDisplay,
    PrecisionRecallDisplay
)

print("Machine learning libraries imported successfully!")
```

```python
# 5.2 Define Features and Target Variable

model_df = eda_df.copy()

X = model_df.drop(columns=["next_semester_dropout"])
y = model_df["next_semester_dropout"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nInput features:")
print(X.columns.tolist())

print("\nTarget distribution:")
print(y.value_counts())
```

```python
# 5.3 Split the Dataset

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
# 5.4 Data Preprocessing

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

numeric_features = X_train.select_dtypes(
    include="number"
).columns.tolist()

categorical_features = X_train.select_dtypes(
    exclude="number"
).columns.tolist()

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"),
         categorical_features)
    ]
)

print("Numerical features:")
print(*numeric_features, sep="\n")

print("\nCategorical features:")
print(*categorical_features, sep="\n")

print("\nPreprocessing configured successfully!")
```

```python
# 5.5 Baseline Model (Dummy Classifier)

dummy_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", DummyClassifier(strategy="most_frequent"))
])

dummy_model.fit(X_train, y_train)

dummy_pred = dummy_model.predict(X_test)

print("Baseline Model Results:")
print("Accuracy:", round(accuracy_score(y_test, dummy_pred), 4))
print("Precision:", round(precision_score(y_test, dummy_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_test, dummy_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_test, dummy_pred, zero_division=0), 4))
```

```python
# 5.6 Logistic Regression Model

from sklearn.base import clone

lr_model = Pipeline([
    ("preprocessor", clone(preprocessor)),
    ("classifier", LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ))
])

lr_model.fit(X_train, y_train)

lr_pred = lr_model.predict(X_test)
lr_prob = lr_model.predict_proba(X_test)[:, 1]

print("Logistic Regression Results:")
print("Accuracy:", round(accuracy_score(y_test, lr_pred), 4))
print("Precision:", round(precision_score(y_test, lr_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_test, lr_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_test, lr_pred, zero_division=0), 4))
print("ROC-AUC:", round(roc_auc_score(y_test, lr_prob), 4))
```

```python
# 5.7 Random Forest Model

rf_model = Pipeline([
    ("preprocessor", clone(preprocessor)),
    ("classifier", RandomForestClassifier(
        n_estimators=100,
        class_weight="balanced",
        random_state=42
    ))
])

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)
rf_prob = rf_model.predict_proba(X_test)[:, 1]

print("Random Forest Results:")
print("Accuracy:", round(accuracy_score(y_test, rf_pred), 4))
print("Precision:", round(precision_score(y_test, rf_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_test, rf_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_test, rf_pred, zero_division=0), 4))
print("ROC-AUC:", round(roc_auc_score(y_test, rf_prob), 4))
```

```python
# 5.8 Model Performance Comparison

dummy_prob = dummy_model.predict_proba(X_test)[:, 1]

models = [
    ("Dummy Classifier", dummy_pred, dummy_prob),
    ("Logistic Regression", lr_pred, lr_prob),
    ("Random Forest", rf_pred, rf_prob)
]

results = []

for name, pred, prob in models:
    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(y_test, pred, zero_division=0),
        "Recall": recall_score(y_test, pred, zero_division=0),
        "F1-score": f1_score(y_test, pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_test, prob)
    })

results_df = pd.DataFrame(results)

display(results_df.round(4))
```

```python
# 5.9 Model Performance Visualization

results_df.set_index("Model")[
    ["Accuracy", "Precision", "Recall", "F1-score", "ROC-AUC"]
].plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Machine Learning Model Performance Comparison")
plt.xlabel("Model")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.legend(loc="upper right")
plt.tight_layout()
plt.show()
```

```python
# 5.10 Confusion Matrices

from sklearn.metrics import ConfusionMatrixDisplay

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

predictions = [
    ("Dummy Classifier", dummy_pred),
    ("Logistic Regression", lr_pred),
    ("Random Forest", rf_pred)
]

for ax, (name, pred) in zip(axes, predictions):
    ConfusionMatrixDisplay.from_predictions(
        y_test,
        pred,
        display_labels=["Continued", "Dropped Out"],
        cmap="Blues",
        ax=ax,
        colorbar=False
    )
    ax.set_title(name)

plt.tight_layout()
plt.show()
```

```python
# 5.11 ROC Curve Comparison

fig, ax = plt.subplots(figsize=(8, 6))

RocCurveDisplay.from_predictions(
    y_test, dummy_prob,
    name="Dummy Classifier", ax=ax
)

RocCurveDisplay.from_predictions(
    y_test, lr_prob,
    name="Logistic Regression", ax=ax
)

RocCurveDisplay.from_predictions(
    y_test, rf_prob,
    name="Random Forest", ax=ax
)

plt.plot([0, 1], [0, 1], "k--", label="Random Guess")
plt.title("ROC Curve Comparison")
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()
```

```python
# 5.12 Precision-Recall Curve Comparison

fig, ax = plt.subplots(figsize=(8, 6))

PrecisionRecallDisplay.from_predictions(
    y_test, dummy_prob,
    name="Dummy Classifier", ax=ax
)

PrecisionRecallDisplay.from_predictions(
    y_test, lr_prob,
    name="Logistic Regression", ax=ax
)

PrecisionRecallDisplay.from_predictions(
    y_test, rf_prob,
    name="Random Forest", ax=ax
)

plt.axhline(
    y=y_test.mean(),
    linestyle="--",
    color="black",
    label="Baseline"
)

plt.title("Precision-Recall Curve Comparison")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# 5.13 Classification Reports

for name, pred in predictions:
    print("\n" + "=" * 50)
    print(name)
    print("=" * 50)

    print(classification_report(
        y_test,
        pred,
        target_names=["Continued", "Dropped Out"],
        zero_division=0
    ))
```
