```python
import pandas as pd

# Load our student retention dataset
df = pd.read_csv("/university_student_retention_dataset_2134.csv")

# Show the first five student records
print("First five rows:")
display(df.head())

# Check the dataset size
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# Check column names
print("Dataset columns:")
print(df.columns.tolist())

# Check the data types
print("Dataset information:")
df.info()

# Check missing values
print("Missing values in each column:")
print(df.isnull().sum())

# Check total missing values
print("Total missing values:", df.isnull().sum().sum())

# Check duplicate rows
print("Duplicate records:", df.duplicated().sum())

# Check the dropout target
print("Student dropout distribution:")
print(df["next_semester_dropout"].value_counts())
```

```python
# STAGE 3 - DATA UNDERSTANDING AND CLEANING

# We already checked duplicate rows in Stage 2.
# Now we check if any student IDs are repeated.

print("Duplicate student IDs:",
      df["student_id"].duplicated().sum())
```

```python
# Let's check if our numerical values make sense.

numeric_columns = [
    "age",
    "gpa_current",
    "gpa_history",
    "attendance_rate",
    "enrollment_gap_months",
    "credits_completed",
    "academic_warning_count",
    "advisor_meeting_count"
]

# Look at the minimum, maximum, and average values.
display(df[numeric_columns].describe())

# Check GPA and attendance.
print("Invalid current GPA:",
      (~df["gpa_current"].between(0, 4)).sum())

print("Invalid previous GPA:",
      (~df["gpa_history"].between(0, 4)).sum())

print("Invalid attendance:",
      (~df["attendance_rate"].between(0, 1)).sum())

# Check for negative values in count columns.
for column in [
    "enrollment_gap_months",
    "credits_completed",
    "academic_warning_count",
    "advisor_meeting_count"
]:
    print(column, "negative values:",
          (df[column] < 0).sum())
```

```python
# We use the IQR method to find unusual values.
# We will review them instead of removing them automatically.

for column in numeric_columns:

    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)

    iqr = q3 - q1

    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr

    outliers = (
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ).sum()

    print(column, "- possible outliers:", outliers)
```

```python
# First, we make a copy of our original dataset.
# This way, the original data stays unchanged.

clean_df = df.copy()

# Remove columns that are not suitable for prediction.
# Some contain personal information or possible data leakage.

columns_to_remove = [
    "student_id",
    "dropout_risk",
    "dropout_type",
    "dropout_duration",
    "dropout_cause",
    "counseling_notes",
    "advising_notes",
    "gpa_by_semester",
    "attendance_by_semester",
    "gender",
    "ethnicity"
]

clean_df = clean_df.drop(columns=columns_to_remove)

print("Remaining columns:")
print(clean_df.columns.tolist())
```

```python
# We create a new column to see how a student's GPA changed.

clean_df["gpa_change"] = (
    clean_df["gpa_current"] - clean_df["gpa_history"]
)

# Let's check our new column.
display(
    clean_df[["gpa_current", "gpa_history", "gpa_change"]].head()
)
```

```python
# Change financial aid status from True/False to 1/0.
clean_df["financial_aid_status"] = (
    clean_df["financial_aid_status"].astype(int)
)

# Make sure the dropout target is numerical.
# 1 = Dropped out, 0 = Continued studying.

clean_df["next_semester_dropout"] = (
    clean_df["next_semester_dropout"].astype(int)
)

print("Updated data types:")
print(clean_df.dtypes)
```

```python
# Let's check the dataset after our changes.

print("Cleaned dataset size:", clean_df.shape)

print("Missing values left:",
      clean_df.isnull().sum().sum())

print("Duplicate rows left:",
      clean_df.duplicated().sum())

# Look at the first five rows.
display(clean_df.head())
```

```python
# Save our cleaned dataset as a new CSV file.

clean_df.to_csv("student_retention_cleaned.csv", index=False)

print("Our cleaned dataset has been saved!")
```

```python
import matplotlib.pyplot as plt

# Chart 1: Student dropout distribution

clean_df["next_semester_dropout"].value_counts().sort_index().plot(
    kind="bar"
)

plt.title("Student Dropout Distribution")
plt.xlabel("Student Status")
plt.ylabel("Number of Students")
plt.xticks([0, 1], ["Continued", "Dropped Out"], rotation=0)
plt.tight_layout()
plt.show()


# Chart 2: GPA and student dropout

clean_df.boxplot(
    column="gpa_current",
    by="next_semester_dropout"
)

plt.title("GPA and Student Dropout")
plt.suptitle("")
plt.xlabel("Dropout Status (0 = No, 1 = Yes)")
plt.ylabel("Current GPA")
plt.tight_layout()
plt.show()


# Chart 3: Attendance and student dropout

clean_df.boxplot(
    column="attendance_rate",
    by="next_semester_dropout"
)

plt.title("Attendance and Student Dropout")
plt.suptitle("")
plt.xlabel("Dropout Status (0 = No, 1 = Yes)")
plt.ylabel("Attendance Rate")
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# STAGE 3 - DATA UNDERSTANDING AND CLEANING
# ==========================================

# We already loaded our dataset as df in Stage 2.
# Now we will check and prepare it for machine learning.


# STEP 1: CHECK DUPLICATES

# Check if any student IDs are repeated
print("Duplicate student IDs:", df["student_id"].duplicated().sum())

# We also check duplicate rows again
print("Duplicate rows:", df.duplicated().sum())


# STEP 2: CHECK MISSING VALUES

# Check which columns have missing information
print("\nMissing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())


# STEP 3: CHECK INCORRECT VALUES

# These are the numerical columns we want to check
numeric_columns = [
    "age",
    "gpa_current",
    "gpa_history",
    "attendance_rate",
    "enrollment_gap_months",
    "credits_completed",
    "academic_warning_count",
    "advisor_meeting_count"
]

# Look at the minimum, maximum, and average values
print("\nSummary of numerical columns:")
display(df[numeric_columns].describe())

# Check if GPA and attendance values are reasonable
print("\nInvalid current GPA:",
      (~df["gpa_current"].between(0, 4)).sum())

print("Invalid previous GPA:",
      (~df["gpa_history"].between(0, 4)).sum())

print("Invalid attendance:",
      (~df["attendance_rate"].between(0, 1)).sum())

# Check for negative numbers where they should not appear
for column in [
    "enrollment_gap_months",
    "credits_completed",
    "academic_warning_count",
    "advisor_meeting_count"
]:
    print("Negative values in", column, ":",
          (df[column] < 0).sum())


# STEP 4: CHECK OUTLIERS

# We use the IQR method to find unusual values
# We will not remove them automatically

print("\nPossible outliers:")

for column in numeric_columns:

    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)

    iqr = q3 - q1

    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr

    outliers = (
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ).sum()

    print(column, ":", outliers)


# STEP 5: REMOVE UNNECESSARY COLUMNS

# Make a copy so our original dataset stays unchanged
clean_df = df.copy()

# Remove columns that are not suitable for prediction
columns_to_remove = [
    "student_id",
    "dropout_risk",
    "dropout_type",
    "dropout_duration",
    "dropout_cause",
    "counseling_notes",
    "advising_notes",
    "gpa_by_semester",
    "attendance_by_semester",
    "gender",
    "ethnicity"
]

clean_df = clean_df.drop(columns=columns_to_remove)

print("\nRemaining columns:")
print(clean_df.columns.tolist())


# STEP 6: CREATE A NEW FEATURE

# This column shows how much a student's GPA changed
clean_df["gpa_change"] = (
    clean_df["gpa_current"] - clean_df["gpa_history"]
)

print("\nGPA change:")
display(clean_df[[
    "gpa_current", "gpa_history", "gpa_change"
]].head())


# STEP 7: CONVERT VALUES INTO NUMBERS

# Change financial aid status from True/False to 1/0
clean_df["financial_aid_status"] = (
    clean_df["financial_aid_status"].astype(int)
)

# Make sure our target is numerical
# 0 = Continued studying, 1 = Dropped out
clean_df["next_semester_dropout"] = (
    clean_df["next_semester_dropout"].astype(int)
)

print("\nUpdated data types:")
print(clean_df.dtypes)


# STEP 8: CHECK OUR CLEANED DATASET

print("\nCleaned dataset size:", clean_df.shape)

print("Missing values left:",
      clean_df.isnull().sum().sum())

print("Duplicate rows left:",
      clean_df.duplicated().sum())

print("\nFirst five rows of our cleaned dataset:")
display(clean_df.head())


# STEP 9: SAVE THE CLEANED DATASET

clean_df.to_csv("student_retention_cleaned.csv", index=False)

print("\nOur cleaned dataset has been saved!")


# STEP 10: CREATE CHARTS

import matplotlib.pyplot as plt

# Chart 1: Student dropout distribution
clean_df["next_semester_dropout"].value_counts().sort_index().plot(
    kind="bar"
)

plt.title("Student Dropout Distribution")
plt.xlabel("Student Status")
plt.ylabel("Number of Students")
plt.xticks([0, 1], ["Continued", "Dropped Out"], rotation=0)
plt.tight_layout()
plt.show()


# Chart 2: GPA and student dropout
clean_df.boxplot(
    column="gpa_current",
    by="next_semester_dropout"
)

plt.title("GPA and Student Dropout")
plt.suptitle("")
plt.xlabel("Dropout Status (0 = No, 1 = Yes)")
plt.ylabel("Current GPA")
plt.tight_layout()
plt.show()


# Chart 3: Attendance and student dropout
clean_df.boxplot(
    column="attendance_rate",
    by="next_semester_dropout"
)

plt.title("Attendance and Student Dropout")
plt.suptitle("")
plt.xlabel("Dropout Status (0 = No, 1 = Yes)")
plt.ylabel("Attendance Rate")
plt.tight_layout()
plt.show()

print("\nStage 3 completed!")
```
